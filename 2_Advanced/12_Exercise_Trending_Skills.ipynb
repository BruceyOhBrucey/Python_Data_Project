```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python

df.info()
```

```python
#Only looking at Data Analyst jobs
df_DA = df[df['job_title_short'] == 'Data Analyst'].copy()
df_DA
```

```python
#add in column showing months (in number)
#Explode job_skills column
df_DA['job_posted_month_no'] = df_DA['job_posted_date'].dt.month
df_DA_explode = df_DA.explode("job_skills")
df_DA_explode
```

```python
#pivot table, with using fill_value = 0 to fill the NaN values to zero
df_DA_pivot = df_DA_explode.pivot_table(
         # The data to aggregate
    index='job_posted_month_no',             # Rows (index of pivot table)
    columns='job_skills',       # Columns (pivot table headers)
    aggfunc='size',                 # Aggregation function
    fill_value=0
)

df_DA_pivot
```

```python
#Create a row at the botom to show the total sum
df_DA_pivot.loc["Total"] = df_DA_pivot.sum()
#arrange by total count and descending order
df_DA_pivot = df_DA_pivot[df_DA_pivot.loc["Total"].sort_values(ascending=False).index] 
#Now we drop the total row
df_DA_pivot = df_DA_pivot.drop("Total")
df_DA_pivot
```

```python
# code snippet from `6_Exercise_Job_Demand.ipynb`
df_DA_pivot = df_DA_pivot.reset_index()
df_DA_pivot['job_posted_month'] = df_DA_pivot['job_posted_month_no'].apply(lambda x: pd.to_datetime(x, format='%m').strftime('%b'))
df_DA_pivot = df_DA_pivot.set_index('job_posted_month')
df_DA_pivot = df_DA_pivot.drop(columns='job_posted_month_no')

df_DA_pivot
```

```python
df_DA_pivot.iloc[:, :5].plot(kind="line")
plt.ylabel("Count")
plt.xlabel("")
plt.title("Top 5 skills for Data Analyst by Month")
plt.show()
```
