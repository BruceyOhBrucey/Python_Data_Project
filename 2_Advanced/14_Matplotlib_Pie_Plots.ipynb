```python
# Importing Libraries
import ast
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
df['job_skills'] = df['job_skills'].apply(lambda x: ast.literal_eval(x) if pd.notna(x) else x)
```

```python
#start the pie chart at 12 oclock with start angle
#Using autopct to show percentage (2 % at the back to showw % in the chart, .1 after %1 to have only 1 decimal place in the chart)
df["job_work_from_home"].value_counts().plot(kind="pie", startangle = 90, autopct="%1.1f%%")
plt.title("Work from Home Status")
plt.ylabel("")
plt.show()

```

```python
df["job_title_short"].value_counts().plot(kind="barh")
#Using GCA invert the values on y-axis
plt.gca().invert_yaxis()
plt.title("Job Titles Count")
plt.ylabel("")
plt.show()
```

```python
fig, ax = plt.subplots(1,3)

#dict_column - key is the column name, values is the title we want for the key (column)

dict_column={
    "job_work_from_home": 'Work from Home',
    "job_no_degree_mention": 'Job Degree Requirements',
    "job_health_insurance": 'Health Insurance offered'
}
 
#for loop just to show what we are looping through
#for i, (column, title) in enumerate(dict_column.items()):
    #print(i, column, title)

#For Loop to plot pie chart for the 3 columns, using () to be a tuple for (column, title)
for i, (column, title) in enumerate(dict_column.items()):
    ax[i].pie(df[column].value_counts(), startangle = 90, autopct="%1.1f%%", labels = ['False', 'True'])
    #set title for each chart
    ax[i].set_title(title, fontsize = 10)

plt.show()
```
