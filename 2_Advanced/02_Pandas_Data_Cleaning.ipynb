```python
# Importing Libraries
import pandas as pd
from datasets import load_dataset
import matplotlib.pyplot as plt  

# Loading Data
dataset = load_dataset('lukebarousse/data_jobs')
df = dataset['train'].to_pandas()

# Data Cleanup
df['job_posted_date'] = pd.to_datetime(df['job_posted_date'])
```

```python
df.info()
```

```python
salaryyear_avg_median = df["salary_year_avg"].median()
salaryhour_avg_median = df["salary_hour_avg"].median()
```

```python
df_filled = df
df_filled["salary_year_avg"] = df_filled["salary_year_avg"].fillna(salaryyear_avg_median)
```

```python
df_filled["salary_hour_avg"] = df_filled["salary_hour_avg"].fillna(salaryhour_avg_median)
```

```python
df_unique = df_filled
df_unique = df_unique.drop_duplicates()

print(len(df_filled))
print(len(df_unique))
```

```python
df_unique = df_unique.drop_duplicates(subset=["job_title", "company_name"])
print(len(df_filled))
print(len(df_unique))
print(len(df_filled)-len(df_unique))
```
